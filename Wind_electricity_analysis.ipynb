If missing values occur mainly before the development of the wind power industry and are followed by low initial values and a gradual increase in generation, treating some of these missing values as zero may be reasonable.

On the other hand, missing values occurring between years with substantial wind generation should not be automatically replaced with zero, as they may represent missing observations rather than actual zero generation.

Further data inspection and cleaning will therefore be performed using SQL.

#Identifying the first observed wind generation

```python
con.execute("""
CREATE OR REPLACE TABLE first_wind AS
SELECT
    country,
    year AS first_wind_year,
    iso_code,
    wind_electricity AS first_wind_value
FROM (
    SELECT
        country,
        year,
        iso_code,
        wind_electricity,
        ROW_NUMBER() OVER (
            PARTITION BY country
            ORDER BY year
        ) AS rn
    FROM df
    WHERE wind_electricity IS NOT NULL
      AND wind_electricity > 0
)
WHERE rn = 1;
""").fetchdf()

first_wind = con.execute("""
    SELECT *
    FROM first_wind
""").fetchdf()

display(first_wind)

```

The first_wind table contains the first observed positive wind generation value for each country. It is used as an intermediate table for investigating whether missing values before the first observation can reasonably be interpreted as zero generation.

The ROW_NUMBER() window function assigns a sequential number to observations within each country after sorting them by year. This makes it possible to select the first positive observation for each country.

# Comparing the first observation with global wind generation

The first observed value is compared with total wind generation reported for the world in the same year.

```python
con.execute("""
CREATE OR REPLACE TABLE wind_first_share AS



SELECT
    f.country,
    f.iso_code,
    f.first_wind_year AS year,
    f.first_wind_value AS first_wind,
    w.world_wind,
    100.0 * f.first_wind_value / w.world_wind AS share
FROM first_wind f
JOIN (
    SELECT
        year,
        SUM(wind_electricity) AS world_wind
    FROM df
    WHERE iso_code IS NOT NULL
    GROUP BY year
) w
ON f.first_wind_year = w.year;
""")
display(con.execute("""
SELECT *
FROM wind_first_share
WHERE iso_code IS NOT NULL
ORDER BY share DESC
""").fetchdf())
```

The world_share value represents the country's first observed positive wind generation as a percentage of total wind generation reported for the same year.

This comparison is used as an indicator of how significant the first observed value is relative to global wind generation.

#Grouping countries by the significance of the first observation

To make the distribution easier to inspect, countries are grouped according to their share of global wind generation.

```python
share_grad = con.execute("""
SELECT
    CASE
        WHEN world_share < 0.01 THEN '<0.01%'
        WHEN world_share < 0.1 THEN '0.01–0.1%'
        WHEN world_share < 1 THEN '0.1–1%'
        WHEN world_share < 5 THEN '1–5%'
        WHEN world_share < 10 THEN '5–10%'
        ELSE '>10%'
    END AS share_group,
    COUNT(*) AS countries
FROM  wind_first_share
GROUP BY share_group
ORDER BY share_group;
""").fetchdf()
display(share_grad)
```

#Checking unusually large shares

The results are checked for cases where a country's first observed value accounts for a very large proportion of global generation.

```python
year_nan = con.execute("""
SELECT country,
       wind_electricity,
       SUM(wind_electricity) OVER () AS world_wind
FROM df WHERE iso_code IS NOT NULL
AND year = 1978
""").fetchdf()
display(year_nan)
```

Most observations without an ISO code correspond to regional or aggregate entities. However, several historical entities also appear without an ISO code and require separate treatment

#Checking countries with no wind generation observations

The data are further examined to distinguish between countries with no reported wind data and countries whose available observations contain only zero values.

```python

display(con.execute("""
SELECT
    country,
    COUNT(*) AS rows,
    COUNT(wind_electricity) AS non_null,
    COUNT(*) FILTER (
        wind_electricity > 0
    ) AS positive,
    COUNT(*) FILTER (
        wind_electricity = 0
    ) AS zero,
    MAX(wind_electricity) AS max_wind
FROM df
WHERE country IN (
    'Kosovo',
    'Czechoslovakia',
    'East Germany',
    'West Germany',
    'USSR',
    'Yugoslavia',
    'Serbia and Montenegro'
)
GROUP BY country
ORDER BY country
""").fetchdf())
```

This check shows that the absence of positive observations can have different meanings: some countries have explicit zero values, while others have missing wind generation data altogether.

Therefore, missing values should not be treated as zero solely on the basis of a country's lack of positive observations.

#Selecting countries for further treatment

Based on the previous checks, countries are separated into groups according to the available information.

Countries with a very small first observed value are considered candidates for treating missing values as zero, while countries with a more substantial first observation require additional attention.

```python
countries_to_zero = con.execute("""

SELECT
    country,
    iso_code,
    year AS first_wind_year,
    first_wind,
    world_wind,
    share
FROM wind_first_share
WHERE share < 0.1
  AND iso_code IS NOT NULL
ORDER BY share

""").fetchdf()

display(countries_to_zero)
```

Countries with a significant first observed value, as well as countries with no wind generation observations, are collected separately for further research.

```python
countries_to_research = con.execute("""
SELECT
    country,
    iso_code,
    year AS first_wind_year,
    first_wind,
    world_wind,
    share,
    'significant_first_wind' AS reason
FROM wind_first_share
WHERE share >= 0.1
  AND iso_code IS NOT NULL

UNION ALL

SELECT
    country,
    iso_code,
    NULL AS first_wind_year,
    NULL AS first_wind,
    NULL AS world_wind,
    NULL AS share,
    'no_data' AS reason
FROM df
WHERE iso_code IS NOT NULL
GROUP BY country, iso_code
HAVING COUNT(wind_electricity) = 0

ORDER BY country
""").fetchdf()

display(countries_to_research)
```

#Additional validation of country groups

Before creating the final dataset, the number and composition of the country groups are checked.

```python
all_with_iso = con.execute("""
SELECT DISTINCT
    country,
    iso_code,
    sum(wind_electricity)
FROM df
WHERE iso_code IS NOT NULL
GROUP BY country, iso_code
HAVING COUNT(wind_electricity) > 0
   AND MAX(wind_electricity) = 0
ORDER BY country
""").fetchdf()

display(all_with_iso)
```

The number of countries without any wind generation observations is also checked.

```python
no_wind_data_count = con.execute("""
SELECT COUNT(*) AS countries_no_wind_data
FROM (
    SELECT
        country,
        iso_code
    FROM df
    WHERE iso_code IS NOT NULL
    GROUP BY country, iso_code
    HAVING COUNT(wind_electricity) = 0
)
""").fetchdf()

display(no_wind_data_count)
```

Finally, the overall distribution of countries is checked.

```python
display(con.execute("""
SELECT
    COUNT(*) AS total_countries,

    COUNT(*) FILTER (
        non_null = 0
    ) AS no_data,

    COUNT(*) FILTER (
        non_null > 0
        AND max_wind = 0
    ) AS zero_only,

    COUNT(*) FILTER (
        max_wind > 0
    ) AS has_positive_wind

FROM (
    SELECT
        country,
        iso_code,
        COUNT(wind_electricity) AS non_null,
        MAX(wind_electricity) AS max_wind
    FROM df
    WHERE iso_code IS NOT NULL
    GROUP BY country, iso_code
) t
""").fetchdf())
```

This provides a final check of the three main categories of countries before the cleaned dataset is created.

#Creating the cleaned wind dataset

After the country-level checks are complete, the variables required for further analysis are selected.

```python
con.execute("""
CREATE OR REPLACE TABLE wind_data AS

SELECT
    country,
    iso_code,
    gdp,
    population,
    year,
    wind_electricity
FROM df
WHERE iso_code IS NOT NULL
""")
```

Only observations with an ISO code are retained in the working dataset.

The countries selected for zero replacement are then registered as a temporary DuckDB table.

```python
con.register("countries_to_zero_df", countries_to_zero)
```

Missing wind generation values for these countries are replaced with zero.

```python
con.execute("""
UPDATE wind_data
SET wind_electricity = 0
WHERE country IN (
    SELECT country
    FROM countries_to_zero_df
)
AND wind_electricity IS NULL
""")
```

DuckDB's UPDATE statement modifies only the rows satisfying the specified conditions, while other columns and rows remain unchanged.

#Validating the cleaned dataset

The number of total, non-null, and remaining null wind generation values is checked for each country.

```python
display(con.execute("""
SELECT
    country,
    iso_code,
    gdp,
    population,
    COUNT(*) AS rows_total,
    COUNT(wind_electricity) AS values_after_cleaning,
    COUNT(*) - COUNT(wind_electricity) AS remaining_null
FROM wind_data
GROUP BY country, iso_code, gdp, population
ORDER BY country
""").fetchdf())
```

The remaining missing values are then examined separately.

```python
display(con.execute("""
SELECT
    country,
    iso_code,
    COUNT(*) AS total_rows,
    COUNT(wind_electricity) AS non_null,
    COUNT(*) - COUNT(wind_electricity) AS null_rows
FROM wind_data
GROUP BY country, iso_code
HAVING (country, iso_code) IN (
    SELECT country, iso_code
    FROM countries_to_zero_df
)
ORDER BY country
""").fetchdf())
```

#Basic validation of population and GDP

Finally, the variables required for subsequent analysis are checked for missing or non-positive values.

```python
display(con.execute("""
SELECT
    country,
    COUNT(*) AS rows,
    COUNT(population) AS non_null,
    MIN(population) AS min_population,
    MAX(population) AS max_population
FROM wind_data
GROUP BY country
ORDER BY non_null;
""").fetchdf())
```

```python
display(con.execute("""
SELECT *
FROM wind_data
WHERE population IS NULL
   OR population <= 0;
""").fetchdf())
```

The same check is performed for GDP.

```python
display(con.execute("""
SELECT
    country,
    COUNT(*) AS rows,
    COUNT(gdp) AS non_null,
    MIN(gdp) AS min_gdp,
    MAX(gdp) AS max_gdp
FROM wind_data
GROUP BY country
ORDER BY non_null;
""").fetchdf())
```

```python
display(con.execute("""
SELECT *
FROM wind_data
WHERE gdp IS NULL
   OR gdp <= 0;
""").fetchdf())
```

```python
factors = [
    "gdp",
    "hdi",
    "population",
    "urbanization",
    "density",
    "network",
    "corruption"
]

target = "wind_electricity"
```

```python
def correlation_analysis(df, target, factors):

    results = []

    for factor in factors:

        data = df[[target, factor]].dropna()

        pearson_r, pearson_p = pearsonr(
            data[target],
            data[factor]
        )

        spearman_r, spearman_p = spearmanr(
            data[target],
            data[factor]
        )

        results.append({
            "factor": factor,
            "n": len(data),
            "pearson_r": pearson_r,
            "pearson_p": pearson_p,
            "spearman_r": spearman_r,
            "spearman_p": spearman_p
        })

    return pd.DataFrame(results)
```

```python
correlation_results = correlation_analysis(
    df,
    target="wind_electricity",
    factors=factors
)

correlation_results
```
